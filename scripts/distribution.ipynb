```python
from qrels import get_correct_uris
import matplotlib.pyplot as plt
from collections import Counter
```

```python
values: dict[int, set[str]] = {}
for i in range(30): values[i] = get_correct_uris(i)
counts: list[int] = [len(v) for v in values.values()]
```

```python
distribution = Counter(counts)
print(distribution)
```

```python
x = sorted(distribution.keys())
y = [distribution[k] for k in x]

positions = range(len(x))

plt.figure(figsize=(15,5))
plt.plot(positions, y, marker="o")

plt.xticks(positions, [str(k) for k in x])

plt.xlabel("Number of relevant documents")
plt.ylabel("Number of INs")

plt.grid(True, alpha=0.3)
plt.show()
```
